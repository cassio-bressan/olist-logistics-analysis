### **Granularidade da base**

A view `vw_olist_full` tem uma linha por combinação **item × pagamento × avaliação**. Um mesmo pedido pode aparecer várias vezes, então médias calculadas direto sobre as linhas dariam mais peso a pedidos com mais itens ou pagamentos. Por isso, todas as métricas de entrega e satisfação são calculadas em **nível de pedido**.

```python
# Linhas vs pedidos únicos
print("Linhas:", len(df))
print("Pedidos únicos:", df["order_id"].nunique())
print("Pedidos com mais de uma avaliação:", (df.groupby("order_id")["review_id"].nunique() > 1).sum())
```

```python
# Corrigindo os tipos de dados de colunas de data
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "shipping_limit_date",
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
# Renomeando colunas para nomes mais curtos e intuitivos
df = df.rename(columns={
    "order_purchase_timestamp": "order_purchase_ts",
    "order_approved_at": "order_approved_ts",
    "order_delivered_carrier_date": "carrier_delivery_ts",
    "order_delivered_customer_date": "customer_delivery_ts",
    "order_estimated_delivery_date": "estimated_delivery_ts",
    "customer_zip_code_prefix": "customer_zip_prefix",
    "price": "item_price",
    "product_category_name": "product_category",
    "seller_zip_code_prefix": "seller_zip_prefix",
    "review_comment_title": "review_title",
    "review_comment_message": "review_message",
    "review_creation_date": "review_creation_ts",
    "review_answer_timestamp": "review_answer_ts"
    })
```

### **Variáveis de entrega**

- `tempo_entrega_real`: dias entre a **compra** e a entrega ao cliente.
- `tempo_entrega_estimado`: dias entre a compra e a data estimada informada ao cliente.
- `dias_atraso`: dias entre a data estimada e a entrega (negativo = entregue antes do prazo).
- `entrega_atrasada`: 1 quando a entrega ocorreu **depois do dia** estimado. A data estimada não tem horário, então uma entrega no próprio dia estimado conta como no prazo.
- `folga_estimativa`: `tempo_entrega_estimado − tempo_entrega_real` (positivo = prazo informado maior que o realizado).

As variáveis só são definidas para pedidos **entregues** (`order_status == "delivered"` com data de entrega). Pedidos cancelados, indisponíveis ou em trânsito ficam com valor vazio e não entram no grupo "No prazo".

```python
# Criando as variáveis de entrega
entregue = (df["order_status"] == "delivered") & df["customer_delivery_ts"].notna()
data_entrega = df["customer_delivery_ts"].dt.normalize()
compra = df["order_purchase_ts"].dt.normalize()

df["pedido_entregue"] = entregue.astype(int)
df["tempo_entrega_real"] = (data_entrega - compra).dt.days.where(entregue)
df["tempo_entrega_estimado"] = (df["estimated_delivery_ts"] - compra).dt.days
df["dias_atraso"] = (data_entrega - df["estimated_delivery_ts"]).dt.days.where(entregue)
df["entrega_atrasada"] = (df["dias_atraso"] > 0).astype("Int64").where(entregue)
df["folga_estimativa"] = df["tempo_entrega_estimado"] - df["tempo_entrega_real"]
```

```python
# Base em nível de pedido: apenas pedidos entregues, uma linha por pedido.
# Quando o pedido tem mais de uma avaliação, fica a mais recente.
pedidos = (
    df[df["pedido_entregue"] == 1]
    .sort_values("review_creation_ts")
    .drop_duplicates("order_id", keep="last")
    .reset_index(drop=True)
)

print("Pedidos entregues analisados:", len(pedidos))
pedidos[["tempo_entrega_real", "tempo_entrega_estimado", "dias_atraso", "entrega_atrasada", "folga_estimativa"]].describe()
```

### **Métricas Principais (KPIs)**

```python
atrasados = pedidos[pedidos["entrega_atrasada"] == 1]

# --- KPI 1: Taxa de pedidos atrasados ---
taxa_atraso = pedidos["entrega_atrasada"].mean()
# --- KPI 2: Atraso médio em dias (somente pedidos atrasados) ---
atraso_medio = atrasados["dias_atraso"].mean()
# --- KPI 3: Tempo médio de entrega real ---
tempo_real_medio = pedidos["tempo_entrega_real"].mean()
# --- KPI 4: Tempo médio estimado ---
tempo_estimado_medio = pedidos["tempo_entrega_estimado"].mean()
# --- KPI 5: Folga média da estimativa (estimado − real) ---
folga_media = pedidos["folga_estimativa"].mean()
```

```python
# Exibir tudo organizado
kpis = {
    "Pedidos entregues analisados": len(pedidos),
    "Taxa de atraso (%)": round(float(taxa_atraso) * 100, 2),
    "Atraso médio dos pedidos atrasados (dias)": round(float(atraso_medio), 2),
    "Tempo médio de entrega real (dias)": round(float(tempo_real_medio), 2),
    "Tempo médio estimado (dias)": round(float(tempo_estimado_medio), 2),
    "Folga média da estimativa (dias)": round(float(folga_media), 2)
}

kpis
```

### **Análise por Seller**

O atraso de um pedido reflete todo o ciclo (seller + transportadora), então o seller não é um proxy puro da transportadora: aqui ele é tratado como o responsável pelo pedido no marketplace.

Um pedido pode ter itens de mais de um seller; nesse caso ele conta para cada seller envolvido. Para evitar rankings dominados por sellers com 1 ou 2 pedidos, o ranking considera apenas sellers com pelo menos `MIN_PEDIDOS_SELLER` pedidos entregues.

```python
MIN_PEDIDOS_SELLER = 20
MIN_ATRASADOS_SELLER = 5

# Uma linha por (pedido, seller), com as métricas do pedido
seller_pedidos = (
    df[df["pedido_entregue"] == 1]
    .dropna(subset=["seller_id"])
    .drop_duplicates(["order_id", "seller_id"])
    [["order_id", "seller_id", "entrega_atrasada", "dias_atraso"]]
)

seller_metrics = (
    seller_pedidos.groupby("seller_id")
    .agg(
        num_pedidos=("order_id", "count"),
        num_atrasados=("entrega_atrasada", "sum"),
        taxa_atraso=("entrega_atrasada", "mean"),
        atraso_medio_dias=("dias_atraso", lambda s: s[s > 0].mean())
    )
    .reset_index()
)

print("Sellers com ao menos 1 pedido entregue:", len(seller_metrics))
print("Sellers com 100% de atraso:", (seller_metrics["taxa_atraso"] == 1).sum(),
      "| desses, com até 2 pedidos:", ((seller_metrics["taxa_atraso"] == 1) & (seller_metrics["num_pedidos"] <= 2)).sum())
```

```python
# Mantendo apenas sellers com volume suficiente para o ranking
seller_metrics = (
    seller_metrics[seller_metrics["num_pedidos"] >= MIN_PEDIDOS_SELLER]
    .sort_values(["taxa_atraso", "num_pedidos"], ascending=False)
    .reset_index(drop=True)
)

seller_metrics["taxa_atraso"] = seller_metrics["taxa_atraso"].round(4)
seller_metrics["atraso_medio_dias"] = seller_metrics["atraso_medio_dias"].round(2)

print("Sellers no ranking:", len(seller_metrics))
seller_metrics.head(10)
```

```python
# Top 10 por atraso médio (gravidade), exigindo um mínimo de pedidos atrasados
seller_metrics[seller_metrics["num_atrasados"] >= MIN_ATRASADOS_SELLER].sort_values("atraso_medio_dias", ascending=False).head(10)
```

### **Relação entre atraso e review score**

```python
# Score médio por atraso vs não atraso (pedidos entregues com avaliação)
avaliados = pedidos.dropna(subset=["review_score"])
rotulos = {0: "No prazo", 1: "Atrasado"}

score_atraso = (
    avaliados.groupby("entrega_atrasada")
    .agg(review_score=("review_score", "mean"), num_pedidos=("order_id", "count"))
    .reset_index()
)
score_atraso["entrega_atrasada"] = score_atraso["entrega_atrasada"].map(rotulos)
score_atraso["review_score"] = score_atraso["review_score"].round(2)
score_atraso
```

```python
# Distribuição de notas por atraso
dist_scores = avaliados.groupby(["entrega_atrasada", "review_score"])["order_id"].count().reset_index()
dist_scores.rename(columns={"order_id": "qtde"}, inplace=True)
dist_scores
```

```python
# Probabilidade de nota 1 com e sem atraso
prob_nota1 = (
    avaliados.assign(nota1=(avaliados["review_score"] == 1).astype(int))
    .groupby("entrega_atrasada")
    .agg(prob_nota1=("nota1", "mean"), num_pedidos=("order_id", "count"))
    .reset_index()
)
prob_nota1["entrega_atrasada"] = prob_nota1["entrega_atrasada"].map(rotulos)
prob_nota1["prob_nota1"] = prob_nota1["prob_nota1"].round(4)
prob_nota1
```

```python
# Comparações usadas no Insights e no README
score_prazo, score_atrasado = score_atraso.set_index("entrega_atrasada")["review_score"][["No prazo", "Atrasado"]]
p_prazo, p_atrasado = prob_nota1.set_index("entrega_atrasada")["prob_nota1"][["No prazo", "Atrasado"]]

print(f"Queda do score médio: {(1 - score_atrasado / score_prazo) * 100:.1f}%")
print(f"Razão da probabilidade de nota 1 (atrasado / no prazo): {p_atrasado / p_prazo:.1f}x")
```

### **Análise por estado**

```python
# Taxa de atraso, atraso médio (somente atrasados) e tempo médio de entrega por estado
atraso_por_estado = (
    pedidos.groupby("customer_state")
    .agg(
        num_pedidos=("order_id", "count"),
        num_atrasados=("entrega_atrasada", "sum"),
        taxa_atraso=("entrega_atrasada", "mean"),
        atraso_medio_dias=("dias_atraso", lambda s: s[s > 0].mean()),
        tempo_entrega_medio=("tempo_entrega_real", "mean")
    )
    .reset_index()
    .sort_values("taxa_atraso", ascending=False)
)

atraso_por_estado["num_atrasados"] = atraso_por_estado["num_atrasados"].astype(int)
atraso_por_estado["taxa_atraso"] = atraso_por_estado["taxa_atraso"].round(4)
atraso_por_estado["atraso_medio_dias"] = atraso_por_estado["atraso_medio_dias"].round(2)
atraso_por_estado["tempo_entrega_medio"] = atraso_por_estado["tempo_entrega_medio"].round(2)
atraso_por_estado
```

```python
# Ranking de gravidade: apenas estados com um mínimo de pedidos atrasados,
# para que a média não dependa de 2 ou 3 pedidos (caso de AP, AM, AC e RR)
MIN_ATRASADOS_ESTADO = 10

atraso_por_estado[atraso_por_estado["num_atrasados"] >= MIN_ATRASADOS_ESTADO].sort_values("atraso_medio_dias", ascending=False).head(10)
```

### **Possíveis “drivers” do atraso**

Exploração em nível de pedido. O tempo real de entrega não entra na correlação porque o atraso é definido a partir dele.

```python
# Variáveis de produto e frete somadas por pedido
itens = df[df["pedido_entregue"] == 1].drop_duplicates(["order_id", "order_item_id"])
por_pedido = itens.groupby("order_id").agg(
    num_itens=("order_item_id", "count"),
    frete_total=("freight_value", "sum"),
    peso_total_g=("product_weight_g", "sum")
)
pedidos_drivers = pedidos[["order_id", "entrega_atrasada", "tempo_entrega_estimado"]].merge(por_pedido, on="order_id")
pedidos_drivers["entrega_atrasada"] = pedidos_drivers["entrega_atrasada"].astype(float)
```

```python
# Atraso por faixa de peso do pedido
atraso_por_peso = (
    pedidos_drivers.groupby(pd.qcut(pedidos_drivers["peso_total_g"], 6), observed=True)
    .agg(taxa_atraso=("entrega_atrasada", "mean"), num_pedidos=("order_id", "count"))
    .reset_index()
)
atraso_por_peso
```

```python
# Correlação com variáveis contínuas
cols_continuas = ["frete_total", "peso_total_g", "num_itens", "tempo_entrega_estimado"]

correlacoes = pedidos_drivers[cols_continuas + ["entrega_atrasada"]].corr()["entrega_atrasada"].drop("entrega_atrasada").sort_values(ascending=False)
correlacoes
```
