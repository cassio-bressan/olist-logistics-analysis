## 1. Introdução
### Objetivo

Este notebook extrai a view analítica final vw_olist_full do banco de dados MySQL usado no projeto e a exporta como um arquivo CSV para a análise posterior em Python.

A etapa de extração separa a obtenção dos dados do fluxo analítico, permitindo que o notebook principal de análise trabalhe sobre uma base local e reproduzível.

```python
from sqlalchemy import create_engine
import pandas as pd
```

## 2. Conexão com o Banco de Dados

```python
USER = "root"
PASSWORD = "#########"
HOST = "localhost"
PORT = 3306
DATABASE = "olist_db"

engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}")
```

## 3. Extração da View Analítica

```python
# Query para extrair a view final
query = "SELECT * FROM vw_olist_full;"

# Lendo a view para um DataFrame
df = pd.read_sql(query, engine)

# Exibindo as primeiras linhas
print(df.head())
```

## 4. Exportação da Base

```python
output_path = "vw_olist_full.csv"

df.to_csv(output_path, index=False)

print(f"Dataset exported to: {output_path}")
```
